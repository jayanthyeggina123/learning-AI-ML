Merged Dataset Analysis :
 The merged analysis combines customer information, products, orders, order items, reviews, and user events to understand purchasing behavior, product performance, customer demographics, and revenue trends.


### Merge Orders with Users

Orders are merged with the Users dataset using the `user_id` column to associate each order with customer information.

```python
orders_users = pd.merge(
    orders,
    users,
    on="user_id",
    how="left"
)

orders_users.head()
```

### Merge Order Items

The order items dataset is merged with the previous dataset using `order_id` to include purchased products and quantities.

```python
sales = pd.merge(
    orders_users,
    order_items,
    on=["order_id", "user_id"],
    how="left"
)

sales.head()
```

### Merge Products

The Products dataset is merged using `product_id` to include product information such as category, brand, price, and rating.

```python
sales = pd.merge(
    sales,
    products,
    on="product_id",
    how="left"
)

sales.head()
```

### Merge Reviews

The Reviews dataset is merged to associate customer ratings and reviews with purchased products.

```python
sales = pd.merge(
    sales,
    reviews[
        [
            "order_id",
            "product_id",
            "rating",
            "review_text"
        ]
    ],
    on=["order_id", "product_id"],
    how="left"
)

sales.head()
```

### Final Dataset

The merged dataset contains customer information, order details, purchased products, and customer reviews in a single table for analysis.

# Customer Behavior Analysis
### 1.Most Viewed Products
This analysis identifies the products that customers viewed the most. High view counts indicate products that attract customer attention and interest.

use the events dataset because it records customer interactions such as view, cart, wishlist, and purchase.

```python
most_viewed = (
    events[events["event_type"] == "view"]
    .groupby("product_id")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .reset_index(name="views")
)

most_viewed = most_viewed.merge(
    products[["product_id", "product_name"]],
    on="product_id",
    how="left"
)

most_viewed
```

```python
fig = px.bar(
    most_viewed,
    x="product_name",
    y="views",
    color="views",
    text="views",
    title="Top 10 Most Viewed Products"
)

fig.update_layout(
    xaxis_title="Product Name",
    yaxis_title="Number of Views"
)

fig.show()
```

### Business Insight

Products with high view counts are popular among customers and can be highlighted in promotional campaigns. If these products have low sales, pricing, reviews, or product descriptions should be evaluated.

### 2.Most Added-to-Cart Products

This analysis identifies the products that customers add to their shopping carts most frequently.

```python
most_cart = (
    events[events["event_type"] == "cart"]
    .groupby("product_id")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .reset_index(name="cart_count")
)

most_cart = most_cart.merge(
    products[["product_id", "product_name"]],
    on="product_id",
    how="left"
)

most_cart
```

```python
fig = px.bar(
    most_cart,
    x="product_name",
    y="cart_count",
    color="cart_count",
    text="cart_count",
    title="Top 10 Most Added-to-Cart Products"
)

fig.update_layout(
    xaxis_title="Product Name",
    yaxis_title="Cart Count"
)

fig.show()
```

### Business Insight

Products with high cart activity are likely to convert into sales. Businesses should ensure these products remain in stock and consider offering discounts or bundle deals to encourage purchases.

### 3 Products with High Views but Low Purchases


This analysis compares product views with purchase events to identify products that attract customer interest but have relatively low purchase activity.

```python
views = (
    events[events["event_type"] == "view"]
    .groupby("product_id")
    .size()
    .reset_index(name="views")
)

purchases = (
    events[events["event_type"] == "purchase"]
    .groupby("product_id")
    .size()
    .reset_index(name="purchases")
)

conversion = views.merge(
    purchases,
    on="product_id",
    how="left"
)

conversion["purchases"] = conversion["purchases"].fillna(0)

conversion["conversion_rate"] = (
    conversion["purchases"] / conversion["views"] * 100
).round(2)

conversion = conversion.merge(
    products[["product_id", "product_name"]],
    on="product_id",
    how="left"
)

high_views_low_purchase = (
    conversion.sort_values(
        by=["views", "conversion_rate"],
        ascending=[False, True]
    )
    .head(10)
)

high_views_low_purchase
```

```python
fig = px.scatter(
    high_views_low_purchase,
    x="views",
    y="conversion_rate",
    size="views",
    hover_name="product_name",
    title="Products with High Views but Low Purchases"
)

fig.update_layout(
    xaxis_title="Views",
    yaxis_title="Conversion Rate (%)"
)

fig.show()
```

### Business Insight

Products with high views but low purchases should be reviewed for pricing, product descriptions, customer reviews, or checkout experience. Improving these factors may increase conversion rates and overall sales.

# Feature engineering 
Feature engineering is the process of creating new features from existing data to improve machine learning models and recommendation systems.

Since the original fashion sales dataset contains only the Clothing category, a new feature called **subcategory** is added to represent different clothing product types. This enables the AI recommendation system to provide more personalized clothing recommendations.

```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
subcategories = [
    "T-Shirt",
    "Shirt",
    "Jeans",
    "Jacket",
    "Hoodie",
    "Dress",
    "Sweater",
    "Shorts",
    "Skirt",
    "Tracksuit"
]
```

```python
weights = [
    0.20,   # T-Shirt
    0.15,   # Shirt
    0.15,   # Jeans
    0.10,   # Jacket
    0.10,   # Hoodie
    0.10,   # Dress
    0.08,   # Sweater
    0.05,   # Shorts
    0.04,   # Skirt
    0.03    # Tracksuit
]
```

```python
fashion_sales["subcategory"] = np.random.choice(
    subcategories,
    size=len(fashion_sales),
    p=weights
)
```
