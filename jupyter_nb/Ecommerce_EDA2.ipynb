# Fashion Sales Dataset – Feature Preparation

The `fashion_sales` dataset was extracted from the original e-commerce dataset and contains sales information related to clothing products.

The original fashion dataset contained **4,605 rows and 22 columns**. During the data preparation and analysis stages, additional features were created to support the business analysis and machine learning recommendation system.

## Additional Features Created

### 1. Order Month

The `order_month` column was created from the `order_date` column.

This feature represents the month in which an order was placed and is useful for analyzing sales and revenue trends over time.

For example:

- `2025-01` → January 2025
- `2025-02` → February 2025
- `2025-03` → March 2025

The `order_month` feature was used for **Monthly Revenue Analysis**.



### 2. Price Range

The `price_range` column was created from the product price to group products into different price categories.

The price ranges help analyze how product prices affect sales and customer purchasing behavior.

The categories include:

- Under $50
- $50-$100
- $100-$200
- $200-$500
- Above $500

The `price_range` feature was used for **Product Sales by Price Range** analysis.



### 3. Subcategory

The original dataset contains only one main category: **Clothing**.

However, the dataset does not contain specific clothing types such as T-Shirts, Jeans, Jackets, or Hoodies. Therefore, a new `subcategory` feature is introduced as an engineered feature to support the personalized fashion recommendation system.

The subcategories include:

- T-Shirt
- Shirt
- Jeans
- Jacket
- Hoodie
- Dress
- Sweater
- Shorts
- Skirt
- Tracksuit

This feature allows the recommendation system to provide more specific clothing recommendations.

## Dataset Evolution

The dataset changed as new features were engineered:

**Original Dataset:**  
4,605 rows × 22 columns

**After adding `order_month` and `price_range`:**  
4,605 rows × 24 columns

**After adding `subcategory`:**  
4,605 rows × 25 columns

The number of rows remains unchanged because these steps only add new features to existing records.

## Purpose of Feature Engineering

Feature engineering helps transform existing data into useful features for analysis and machine learning.

The additional features allow the project to analyze:

- Monthly sales and revenue trends
- Product sales across different price ranges
- Clothing subcategory performance
- Customer preferences
- More personalized product recommendations

These engineered features will later be used in the **AI Personalized Fashion Shopping Assistant**.

### Visualize subcategory distribution

```python
product_type_count = (
    fashion_sales["product_type"]
    .value_counts()
    .reset_index()
)

product_type_count.columns = ["product_type", "count"]

fig = px.bar(
    product_type_count,
    x="product_type",
    y="count",
    color="count",
    title="Distribution of Clothing Product Types",
    labels={
        "product_type": "Product Type",
        "count": "Number of Records"
    }
)

fig.show()
```

# Business Analysis for Subcategory

### Revenue by product_type

```python
product_type_revenue = (
    fashion_sales
    .groupby("product_type")["item_total"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    product_type_revenue,
    x="product_type",
    y="item_total",
    color="item_total",
    title="Revenue by Product Type",
    labels={
        "product_type": "Product Type",
        "item_total": "Revenue ($)"
    }
)

fig.show()
```

### Best-Selling Product Types

```python
product_type_sales = (
    fashion_sales
    .groupby("product_type")["quantity"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    product_type_sales,
    x="product_type",
    y="quantity",
    color="quantity",
    title="Best-Selling Product Types",
    labels={
        "product_type": "Product Type",
        "quantity": "Units Sold"
    }
)

fig.show()
```

### Average Rating by Product Type

```python
product_type_rating = (
    fashion_sales
    .groupby("product_type")["product_rating"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    product_type_rating,
    x="product_type",
    y="product_rating",
    color="product_rating",
    title="Average Product Rating by Product Type",
    labels={
        "product_type": "Product Type",
        "product_rating": "Average Rating"
    }
)

fig.show()
```
