# Text Data

Text data is information in the form of words or sentences.

Examples:

```text
"I love pizza"
"Python is easy"
"This product is very good"
```

Text data is not directly understood by machine learning models.

So we first need to **preprocess and convert text into numerical form**.


# NLP

NLP stands for **Natural Language Processing**.

It is used to help computers work with and understand human language.

Examples:

- Text classification
- Sentiment analysis
- Spam detection
- Chatbots
- Machine translation
- Text generation


# NLU

NLU stands for **Natural Language Understanding**.

It focuses on understanding the **meaning and intent** of human language.

Example:

```text
"Can you book a flight to Boston?"
```

NLU tries to understand:

```text
Intent → Book flight
Destination → Boston
```


# NLP vs NLU

```text
NLP
↓
Works with human language

NLU
↓
Focuses on understanding meaning, intent and context
```


# Text Preprocessing

Before applying an NLP model, text is usually cleaned and processed.

Common preprocessing techniques:

```text
Tokenization
POS Tagging
Chunking
Stemming
Lemmatization
Named Entity Recognition
```

These techniques help prepare text for further processing.


# Tokenization

Tokenization means **breaking text into smaller pieces called tokens**.

Example:

```text
"I love pizza"
```

After tokenization:

```text
["I", "love", "pizza"]
```

Each word becomes a token.

Character/tokenization can also be done at sentence or subword level depending on the model.


# POS Tagging

POS stands for **Part-of-Speech**.

POS tagging identifies the grammatical role of each word.

Example:

```text
"Jayanth eats pizza"
```

```text
Jayanth → Noun
eats    → Verb
pizza   → Noun
```

Common POS tags:

```text
Noun
Verb
Adjective
Adverb
Pronoun
Preposition
```


# Chunking

Chunking groups words into meaningful phrases.

Example:

```text
"The young boy eats pizza"
```

Possible chunks:

```text
[The young boy] → Noun Phrase
[eats]          → Verb
[pizza]         → Noun Phrase
```

Chunking is sometimes called **shallow parsing**.


# Stemming

Stemming reduces a word to a simpler/root-like form by removing prefixes or suffixes.

Example:

```text
playing
played
plays
```

May become:

```text
play
```

Another example:

```text
studies → studi
```

## Drawback of Stemming

Stemming does not always produce a proper English word.

Example:

```text
studies → studi
```

So it can lose some meaning or produce incorrect words.


# Lemmatization

Lemmatization also reduces words to their base form, but it tries to produce a **proper dictionary word**.

Example:

```text
running → run
better  → good
studies → study
```

Lemmatization usually considers the word's meaning and grammatical role.

```text
Stemming
→ Faster
→ Can produce incorrect words

Lemmatization
→ More accurate
→ Usually slower
→ Produces meaningful base words
```


# Named Entity Recognition (NER)

NER stands for **Named Entity Recognition**.

It identifies important entities in text.

Example:

```text
"Jayanth lives in Boston and works at Amazon."
```

NER may identify:

```text
Jayanth → PERSON
Boston  → LOCATION
Amazon  → ORGANIZATION
```

Common entities:

```text
PERSON
LOCATION
ORGANIZATION
DATE
MONEY
PRODUCT
```

## Drawback of Basic NER

Some NER methods may struggle when there is **not enough context** or when a word has multiple meanings.

Example:

```text
"Apple released a new phone."
```

Here:

```text
Apple → Company
```

But:

```text
"I ate an apple."
```

Here:

```text
apple → Fruit
```

The surrounding context is important.


# Types of NER Approaches

## Rule-Based NER

Uses manually created rules and patterns.

Example:

```text
If a word follows "Mr."
→ likely a person's name
```

### Drawback

- Requires many manually written rules.
- Difficult to maintain.
- May fail on new patterns.


## Machine Learning Based NER

Uses labeled training data to learn entity patterns.

```text
Training Data
     ↓
ML Model
     ↓
Learn Entity Patterns
     ↓
NER Prediction
```

### Drawback

- Requires labeled training data.
- Performance depends on the training data.


## Transformer-Based NER

Uses Transformer models to understand context and identify entities.

```text
Text
 ↓
Transformer
 ↓
Context Understanding
 ↓
Entity Detection
```

Generally more powerful than simple rule-based approaches because they can use surrounding context.


# BIO Tagging

BIO tagging is commonly used for **Named Entity Recognition**.

```text
B → Beginning of an entity
I → Inside an entity
O → Outside an entity
```

Example:

```text
"Jayanth lives in New York"
```

BIO tags:

```text
Jayanth → B-PER
lives   → O
in      → O
New     → B-LOC
York    → I-LOC
```

Here:

```text
B-PER → Beginning of Person
B-LOC → Beginning of Location
I-LOC → Inside Location
O     → Not an entity
```


# BIO Tagging Step-by-Step

Sentence:

```text
"Jayanth works at Amazon"
```

### Step 1: Tokenization

```text
Jayanth
works
at
Amazon
```

### Step 2: Identify Entities

```text
Jayanth → PERSON
Amazon  → ORGANIZATION
```

### Step 3: Apply BIO Tags

```text
Jayanth → B-PER
works   → O
at      → O
Amazon  → B-ORG
```

If the entity contains multiple words:

```text
New York
```

Then:

```text
New  → B-LOC
York → I-LOC
```


# N-Gram Model

An N-Gram is a sequence of **N words/tokens**.

Examples:

```text
Unigram → 1 word
Bigram  → 2 words
Trigram → 3 words
```

Sentence:

```text
"I love eating pizza"
```

## Unigram

```text
I
love
eating
pizza
```

## Bigram

```text
I love
love eating
eating pizza
```

## Trigram

```text
I love eating
love eating pizza
```


# Word Vectorization / Word Embedding

Machine learning models cannot directly understand words as text.

So words are converted into **numbers/vectors**.

```text
Word
 ↓
Vectorization
 ↓
Numerical Vector
```

Example:

```text
"king" → [0.25, 0.71, 0.34, ...]
```

This allows ML models to work with text.


# One-Hot Encoding

One-Hot Encoding represents each word using a vector containing mostly zeros and one `1`.

Example:

```text
Words:

cat
dog
fish
```

Representation:

```text
cat  → [1, 0, 0]
dog  → [0, 1, 0]
fish → [0, 0, 1]
```

## Drawback

For a very large vocabulary, vectors become very large and mostly contain zeros.

This is called a **sparse representation**.

```text
Large Vocabulary
      ↓
Very Large Vectors
      ↓
Mostly 0s
      ↓
High memory usage
```


# Bag of Words

Bag of Words represents text based on the **frequency of words**.

Example:

```text
Document 1:
"I love pizza"

Document 2:
"I love Python"
```

Vocabulary:

```text
I
love
pizza
Python
```

Representation:

```text
Document 1 → [1, 1, 1, 0]
Document 2 → [1, 1, 0, 1]
```

The model mainly considers **which words occur and how often**.


# Drawback of Bag of Words

Bag of Words does not properly understand:

- Word order
- Context
- Meaning

Example:

```text
"Dog bites man"
"Man bites dog"
```

The words are similar, but the meaning is different.

Bag of Words may treat them too similarly because it mainly focuses on word frequency.


# TF-IDF

TF-IDF stands for:

**Term Frequency - Inverse Document Frequency**

It gives importance to words based on:

```text
How often the word appears in a document
        +
How rare the word is across documents
```


# Term Frequency (TF)

Term Frequency measures **how frequently a word appears in a document**.

Example:

```text
Document:
"I love pizza. Pizza is my favorite food."

pizza appears 2 times
```

So pizza has a relatively high Term Frequency.


# Inverse Document Frequency (IDF)

IDF measures how **rare or common a word is across multiple documents**.

A word appearing in many documents gets lower importance.

Example:

```text
"the"
```

appears in many documents.

So:

```text
IDF → Low
```

A rare word:

```text
"transformer"
```

may appear in fewer documents.

So:

```text
IDF → Higher
```


# TF-IDF

```text
TF-IDF
   =
Term Frequency
      ×
Inverse Document Frequency
```

Simple idea:

```text
Common word
→ Lower importance

Frequent in one document but rare across documents
→ Higher importance
```


# N-Grams and Frequency

Unigram, Bigram, Trigram and TF-IDF-based approaches mainly depend on **word occurrence/frequency**.

```text
Unigram → Word frequency
Bigram  → Word-pair frequency
Trigram → Three-word sequence frequency
```

They can capture some local word order using bigrams/trigrams, but they still have limitations in understanding **long-range context and deeper meaning**.


# Skip-Gram

Skip-Gram is a word embedding approach.

Instead of using nearby words to predict a target word, it tries to predict surrounding/context words from a target word.

Example:

```text
"The cat is sitting"

Target word → cat

Context words → The, is
```

The model learns relationships between words based on their surrounding context.

Skip-Gram is associated with **Word2Vec**.


# Continuous Bag of Words (CBOW)

CBOW stands for **Continuous Bag of Words**.

It predicts a target word using surrounding/context words.

Example:

```text
"The cat is sitting"
```

If the target word is:

```text
cat
```

The surrounding words may be used as input:

```text
The
is
```

Then:

```text
Context Words
      ↓
   Neural Network
      ↓
Target Word
```

CBOW is also associated with **Word2Vec**.


# CBOW Architecture

Basic structure:

```text
Input Layer
     ↓
Hidden Layer
     ↓
Output Layer
```

Example:

```text
Context Words
      ↓
Input Layer
      ↓
Hidden Layer
      ↓
Output Layer
      ↓
Target Word
```

CBOW uses context better than simple Bag of Words because it learns word relationships from surrounding words.


# Neural Networks for Text

Neural networks can learn numerical representations of words and patterns in text.

```text
Text
 ↓
Tokenization
 ↓
Vectorization / Embedding
 ↓
Neural Network
 ↓
Prediction
```


# RNN

RNN stands for **Recurrent Neural Network**.

RNN is designed to work with sequential data.

It passes information from one step to the next.

```text
Input 1 → Hidden → Output
             ↓
Input 2 → Hidden → Output
             ↓
Input 3 → Hidden → Output
```

The previous hidden state is passed to the next step.

This allows the model to use some previous information.


# RNN Example

Sentence:

```text
"I love eating pizza"
```

The RNN processes the words sequentially:

```text
I
 ↓
love
 ↓
eating
 ↓
pizza
```

Information from earlier words can influence later predictions.


# Drawbacks of RNN

- Processes sequences one step at a time.
- Difficult to process long sequences efficiently.
- Can have difficulty remembering long-term information.
- Training can be slower because of sequential processing.
- Can suffer from vanishing/exploding gradient problems.


# LSTM

LSTM stands for **Long Short-Term Memory**.

LSTM is an improved type of RNN designed to remember important information for longer periods.

It uses gates to control what information should be:

```text
Remembered
Forgotten
Added
```

Simple idea:

```text
Previous Information
        ↓
      LSTM
        ↓
Important Information
        ↓
Next Step
```


# LSTM Gates

LSTM mainly uses:

```text
Forget Gate
Input Gate
Output Gate
```

### Forget Gate

Decides what old information should be removed.

### Input Gate

Decides what new information should be stored.

### Output Gate

Decides what information should be passed to the next step/output.


# LSTM Drawbacks

- Requires more memory than simpler models.
- More complex than basic RNN.
- Sequential processing makes parallelization difficult.
- Training can be slow for very long sequences.


# GRU

GRU stands for **Gated Recurrent Unit**.

GRU is another improved version of RNN.

It uses gates to control information flow but has a simpler structure than LSTM.

```text
RNN
 ↓
GRU
 ↓
Better handling of long-term information
```

GRU generally has fewer gates/parameters than LSTM, so it can be faster and simpler.


# RNN vs LSTM vs GRU

| Model | Main Idea |
|---|---|
| **RNN** | Processes sequential information |
| **LSTM** | Better memory for long-term information |
| **GRU** | Simpler gated version of RNN |


# Common Problems with RNN/LSTM/GRU

```text
Long Context
     ↓
Difficult to remember everything

Serial Processing
     ↓
Cannot process all sequence positions independently

High Memory / Computation
     ↓
Long sequences become expensive

Complex Models
     ↓
More difficult to train and maintain
```


# Sequence-to-Sequence Model

Sequence-to-Sequence models are used when the input and output are both sequences.

Example:

```text
English Sentence
      ↓
   Encoder
      ↓
Representation
      ↓
   Decoder
      ↓
French Sentence
```

Example:

```text
"I love Python"
       ↓
    Encoder
       ↓
   Representation
       ↓
    Decoder
       ↓
"J'aime Python"
```

Common applications:

- Machine translation
- Text summarization
- Chatbots
- Question answering


# Encoder-Decoder

The **Encoder** reads the input sequence and creates a representation.

The **Decoder** uses that representation to generate the output sequence.

```text
Input
 ↓
Encoder
 ↓
Representation
 ↓
Decoder
 ↓
Output
```


# Transformer Model

The Transformer is a neural network architecture designed to process sequences using **attention** instead of relying mainly on recurrence.

A major component is:

```text
Multi-Head Attention
```

Transformers can process sequence positions more **in parallel** than RNN-based models during training.


# Attention

Attention allows the model to determine which words/tokens are important when processing a particular word.

Example:

```text
"The animal didn't cross the road because it was tired."
```

The model can use attention to understand which words are related to:

```text
"it"
```

Attention helps the model understand relationships between words even when they are far apart in a sentence.


# Multi-Head Attention

Instead of using only one attention mechanism, Transformer models use multiple attention heads.

```text
Input
  ↓
 ┌──────┬──────┬──────┐
 ↓      ↓      ↓
Head 1 Head 2 Head 3
 ↓      ↓      ↓
 └──────┴──────┴──────┘
          ↓
      Combined
```

Different attention heads can learn different relationships in the text.


# Transformer vs RNN

```text
RNN
↓
Sequential processing
↓
Difficult to parallelize
↓
Long context can be difficult
```

```text
Transformer
↓
Attention mechanism
↓
More parallel processing
↓
Better handling of long-range relationships
```

This helped Transformers become very important in modern NLP.


# Evolution of NLP Models

A simple progression:

```text
Text
 ↓
Tokenization
 ↓
Stemming / Lemmatization
 ↓
Bag of Words
 ↓
TF-IDF
 ↓
Word Embeddings
 ↓
Word2Vec
 ↓
RNN
 ↓
LSTM / GRU
 ↓
Sequence-to-Sequence
 ↓
Transformer
 ↓
Modern NLP / LLMs
```


# Main Idea to Remember

```text
Traditional NLP
→ Mainly frequency and word occurrence

Word Embeddings
→ Learn relationships between words

RNN / LSTM / GRU
→ Learn sequential information

Transformer
→ Uses attention to understand relationships
   and supports much more parallel processing
```
