# Transformer:


A **Transformer** is a type of neural network architecture designed to process sequences, especially language.

The most important idea behind a Transformer is **attention**.

Transformer = a neural network that uses attention to understand relationships between tokens.


### 2. Tokenization

Tokenization is the process of breaking text into smaller pieces called tokens.

A token can be a complete word, part of a word, punctuation, or another piece of text.

Sentence:

The dog is running.

The | dog | is | running | .

Tokenization = breaking text into pieces.

### 3. Token IDs

A computer cannot directly process words such as "dog" in the same way a human does.

Tokens are therefore associated with numerical IDs.

Conceptually:
"The"      → 125

"dog"      → 732

"is"       → 41

"running"  → 982

These numbers are examples only. Actual token IDs depend on the tokenizer and vocabulary.

Token ID = the number used to identify a token.

### 4. Embeddings

An embedding is a numerical vector that represents a token.

Instead of working directly with the word "dog", the neural network works with a vector of numbers.

Conceptually:
"dog"
  |
  v
[0.21, -0.45, 0.73, 0.18, ...]

The actual embedding contains many more numbers.

The important idea is that the word has been converted into a numerical representation that the neural network can process.

Embedding = representing a token using numbers.

### 5.Positional Information

A Transformer needs information about the position or order of tokens.

This is important because changing the order of words can change the meaning of a sentence.

Compare:

Dog bites man.

with:

Man bites dog.

The same words appear, but the meaning is different.

The Transformer therefore needs to know where each token occurs.

### 6. Attention

Attention allows a Transformer to determine which tokens are important to another token.

Instead of treating every word as completely independent, the model looks at relationships between tokens.

Consider:

The dog ate the food because it was hungry.

What does "it" refer to?

The model can learn a strong relationship between:

"it"  <-------->  "dog"

It may also consider other words such as "food".

The attention mechanism calculates how strongly the tokens should influence one another.

Attention = deciding what information is important.

                    "it"
                      |
                      v
          Which other tokens matter?
                      |
          +-----------+-----------+
          |           |           |
          v           v           v
         dog        food       hungry
          |
          v
   Strong relationship


### 7. Self-Attention

Self-attention means that tokens in the same sequence can attend to other tokens in that sequence.

The model uses the tokens themselves to understand their relationships.


Consider:

The dog is eating because it is hungry.

The model can examine relationships such as:
"it"       <------> "dog"

"hungry"   <------> "dog"

"eating"   <------> "dog"

Each token can consider information from other tokens.

Self-attention = tokens looking at other tokens in the same sequence.

### 9. Attention Scores

An attention score represents how strongly one token should pay attention to another token.

Higher attention means the relationship is considered more important.

Suppose the model processes:

The dog ate the food because it was hungry.

The model might calculate relationships such as:

"it" → "dog"       High attention

"it" → "food"      Lower attention

The exact values are calculated mathematically by the model.

Attention score = how important another token is.

### 11. Multi-Head Attention

Multi-head attention uses multiple attention mechanisms, called heads, at the same time.

Each head can learn to focus on different relationships.

Imagine several people reading the same sentence.

One person might focus on grammar.

Another might focus on relationships between words.

Another might focus on another pattern.

Multi-head attention = multiple ways of looking at relationships.

### 12. Feed-Forward Network

A feed-forward network is a neural-network component that further processes the information after the attention step.

Attention helps determine what information is important.

The feed-forward network processes that information further.

Attention finds important information; feed-forward processing processes it further.

Think of it like:

Attention:
"What information matters?"

        ↓

Feed-Forward:
"Let's process that information."


### 13. Add & Norm

Add & Norm is a part of a Transformer block that combines information through a residual connection and applies layer normalization.

Add helps preserve earlier information.

Norm helps stabilize the values.


The original information can be carried forward while the newly processed information is added.

### 16. Final Representation
After the last Transformer block, we have a final representation.

This is basically the model's processed numerical representation of the information it has understood so far.

Think of it as:

"After all my processing, this is what I currently understand about the input."

For example:

"The dog is"

       ↓

       ↓

Final Representation

[0.12, -0.73, 0.44, 0.91, ...]

Final representation = the information the Transformer has produced after processing the input.

### 17. Linear Layer
Now the model needs to turn that representation into scores for every possible token in its vocabulary.

This is where the linear layer comes in.

Suppose the vocabulary contains:

The linear layer takes the final representation and produces a score for each possible token.

Linear layer = converts the final representation into a score for every possible next token.

### 18. Logits
The scores produced by the linear layer are called logits.


Token        Logit

dog          1.2

cat          0.8

hungry       4.7

running      3.1

car          0.4

house        0.9

These numbers are not probabilities yet.

They are just raw scores.

A higher logit generally means the model considers that token more likely.

Here:

hungry = 4.7

running = 3.1

dog = 1.2

So "hungry" currently has the strongest score.

Logits = raw scores for possible tokens.

### 19. Softmax
Now we have a problem.

The logits aren't easy to interpret as probabilities.

For example:

hungry  = 4.7

running = 3.1

dog     = 1.2

We want something like:

hungry  → 70%

running → 20%

dog     → 5%
...
This is where Softmax comes in.

Softmax converts the raw logits into a probability distribution.

LOGITS
  ↓
Softmax
  ↓
PROBABILITIES

Conceptually:

dog       → 0.05

cat       → 0.02

hungry    → 0.70

running   → 0.20

car       → 0.01

house     → 0.02

The probabilities add up to approximately:

1.00
or:
100%


### 20. Probability
Now the model has a probability for every possible next token.

For example:

"The dog is"

        ↓

Possible next tokens:

hungry      70%

running     20%

sleeping     5%

outside      3%

happy        1%

...          1%

The model is basically saying:

"Based on everything I have processed, I think 'hungry' is the most likely next token."

### 21. Choosing the Next Token
Now the system needs to actually choose a token.

There are different ways to do this.

One simple method is to choose the token with the highest probability.

hungry      70%  ← CHOOSE

running     20%

sleeping     5%

outside      3%

happy        1%

So:

Next Token = "hungry"

However, modern language models can also sample from the probability distribution rather than always choosing the 
highest-probability token.

That's where concepts such as:

temperature

top-k

top-p
can come into play.

### Why Is It Called a "Decoder" If There Is No Encoder?

This is a common point of confusion.

You might think:

"If GPT is decoder-only, what is it decoding?"

The name comes from the original Transformer architecture, which had an encoder and decoder.

Modern decoder-only models use the decoder-style Transformer block, especially its masked/self-attention behavior, even though they don't have a separate encoder.

So:

Original Transformer:

Encoder + Decoder

GPT-style:

Decoder only

### Self-Attention

The model looks at tokens within the same sequence.
Token A
   ↕

It helps understand relationships within the sequence.
