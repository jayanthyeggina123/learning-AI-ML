# Linear Regression 

## Linear Relationships

A linear relationship means two variables have a relationship that can be represented approximately by a straight line.

Example:

```text
Hours Studied ↑ → Exam Marks ↑
```

If study hours increase and marks also increase, there may be a positive linear relationship.

Basic equation:

```text
Y = mx + c
```

- **Y** = Target/Dependent variable
- **X** = Independent variable
- **m** = Slope
- **c** = Intercept


## Positive Correlation

Positive correlation means when one variable increases, the other variable also tends to increase.

Example:

```text
Study Hours ↑ → Marks ↑
```

Another example:

```text
Experience ↑ → Salary ↑
```

Correlation value is between:

```text
-1 to +1
```

A value closer to `+1` means strong positive correlation.


## No Correlation

No correlation means there is no clear linear relationship between two variables.

Example:

```text
Shoe Size ↔ Exam Marks
```

A person's shoe size normally does not tell us how well they will perform in an exam.

```text
Correlation = 0
```

means there is no linear correlation.


## Negative Correlation

Negative correlation means when one variable increases, the other variable tends to decrease.

Example:

```text
Price ↑ → Demand ↓
```

Another example:

```text
Age of Car ↑ → Resale Value ↓
```

A value closer to `-1` means strong negative correlation.

```text
-1 → Perfect negative correlation
```


# Simple Linear Regression

Simple Linear Regression uses **one independent variable** to predict one dependent variable.

Equation:

```text
Y = mx + c
```

Example:

```text
X = Hours Studied
Y = Exam Marks
```

The model learns the relationship between study hours and marks and predicts marks for a new student.


# Pearson Correlation

Pearson Correlation measures the **strength and direction of a linear relationship** between two numerical variables.

The value ranges from:

```text
-1 to +1
```

| Value | Meaning |
|---|---|
| +1 | Perfect positive correlation |
| 0 | No linear correlation |
| -1 | Perfect negative correlation |

Example:

```text
Study Hours and Marks

Pearson r = 0.90
```

This means there is a strong positive linear relationship.

### Important

Correlation does not always mean causation.

Two variables can be related without one directly causing the other.


# Multicollinearity

Multicollinearity occurs in Multiple Linear Regression when **two or more independent variables are highly correlated with each other**.

Example:

```text
House Size
House Area in Square Feet
```

These two features may contain almost the same information.

Because of this, the model may have difficulty determining the individual effect of each feature.

### Problems

- Coefficients can become unstable.
- It becomes difficult to understand the effect of each feature.
- Model reliability can be affected.


# Remedies for Multicollinearity

Some ways to handle multicollinearity:

- Remove one of the highly correlated features.
- Combine related features.
- Use PCA (Principal Component Analysis).
- Use Regularization such as Ridge Regression.


# Principal Component Analysis (PCA)

PCA is a **dimensionality reduction technique**.

It converts many related features into a smaller number of new variables called **Principal Components**.

Example:

```text
Height
Weight
BMI
Body Size
```

PCA can combine the information from these features into fewer components.

### Why use PCA?

- Reduce the number of features.
- Reduce multicollinearity.
- Make data easier to visualize.
- Reduce computational complexity.


# Bias

Bias is the error caused when a model is **too simple** and cannot properly understand the patterns in the data.

High Bias usually causes **Underfitting**.

Example:

A complex curved relationship is modeled using a very simple straight line.

```text
High Bias → Underfitting
```


# Variance

Variance means how much the model's predictions change when it is trained using different training data.

A high-variance model is very sensitive to the training data.

High Variance usually causes **Overfitting**.

```text
High Variance → Overfitting
```


# Total Error

Total error can be represented as:

```text
Total Error = Bias² + Variance + Irreducible Error
```

### Bias²

Error caused by the model being too simple.

### Variance

Error caused by the model being too sensitive to the training data.

### Irreducible Error

Random noise or errors in the data that cannot be completely removed.


# Bias and Variance Combinations

## Low Bias + Low Variance

This is the ideal situation.

```text
Low Bias + Low Variance
        ↓
Good Model
```

The model performs well and generalizes well to new data.


## Low Bias + High Variance

This usually means **Overfitting**.

The model learns the training data too closely.

Example:

```text
Training Accuracy = 99%
Testing Accuracy = 70%
```


## High Bias + Low Variance

This usually means **Underfitting**.

The model is too simple to understand the important patterns.

Example:

```text
Training Accuracy = 70%
Testing Accuracy = 68%
```


## High Bias + High Variance

This is a poor model because it is both inaccurate and unstable.


# Bias-Variance Trade-off

Bias-Variance Trade-off means finding the right balance between **bias and variance**.

Generally:

```text
Simple Model
    ↓
Higher Bias + Lower Variance
```

and

```text
Complex Model
    ↓
Lower Bias + Higher Variance
```

The goal is to find a model that has a good balance between bias and variance.


# Regularization

Regularization is a technique used to **reduce overfitting** by adding a penalty to large model coefficients.

Without regularization:

```text
Complex Model
      ↓
Overfitting
```

With regularization:

```text
Penalty on large coefficients
      ↓
Simpler Model
      ↓
Less Overfitting
```


# Why Use Regularization?

Regularization is used to:

- Reduce overfitting.
- Control model complexity.
- Prevent coefficients from becoming unnecessarily large.
- Improve performance on unseen data.
- Help with multicollinearity.


# Lasso Regression - L1 Regularization

Lasso Regression uses **L1 Regularization**.

Lasso can shrink some coefficients completely to **zero**.

```text
Lasso = L1
```

Example:

```text
Brand  → 2.5
Rating → 1.8
Color  → 0
Size   → 0
Price  → 3.1
```

Here, Lasso made the coefficients of `Color` and `Size` equal to zero.

This means those features can effectively be removed from the model.

### Easy way to remember

```text
Lasso → L1 → Can make coefficients ZERO
```


# Ridge Regression - L2 Regularization

Ridge Regression uses **L2 Regularization**.

Ridge shrinks coefficients toward zero but normally does not make them exactly zero.

Example:

Before Ridge:

```text
Brand  = 5.0
Rating = 4.0
Color  = 3.0
```

After Ridge:

```text
Brand  = 3.8
Rating = 2.9
Color  = 1.7
```

The values become smaller, but the features remain in the model.

### Easy way to remember

```text
Ridge → L2 → Shrinks coefficients
```


# Lasso vs Ridge

| Lasso | Ridge |
|---|---|
| Uses L1 Regularization | Uses L2 Regularization |
| Can make coefficients exactly 0 | Usually does not make coefficients exactly 0 |
| Can perform feature selection | Keeps features but shrinks their coefficients |
| Useful when some features may be unnecessary | Useful when many features contribute to prediction |


# Elastic Net

Elastic Net combines both **L1 and L2 Regularization**.

```text
Elastic Net = L1 + L2
```

It combines some advantages of both Lasso and Ridge.

It can:

- Shrink coefficients.
- Make some coefficients zero.
- Help with multicollinearity.


# Lasso, Ridge and Elastic Net

```text
Lasso       → L1
Ridge       → L2
Elastic Net → L1 + L2
```


# Polynomial Regression

Polynomial Regression is used when the relationship between variables is **curved instead of a straight line**.

Simple Linear Regression:

```text
Y = b0 + b1X
```

Polynomial Regression:

```text
Y = b0 + b1X + b2X² + b3X³ + ... + bnXⁿ
```

It adds powers of X to capture a curved relationship.


# Polynomial Degree

The degree tells us the **highest power of X** in the equation.

## Degree 1

```text
Y = b0 + b1X
```

This is basically Linear Regression.


## Degree 2

```text
Y = b0 + b1X + b2X²
```

This can create a curved relationship.


## Degree 3

```text
Y = b0 + b1X + b2X² + b3X³
```

This creates a more flexible curve.


### Important

Higher degree means the model becomes more flexible.

But too much flexibility can cause:

```text
Higher Degree
     ↓
More Complex Model
     ↓
Higher Risk of Overfitting
```


# Classification or Linear Regression?

It depends on **what we want to predict**.

## Linear Regression

Linear Regression is used when the target is a **continuous numerical value**.

Examples:

```text
House Price = $250,000
Salary = $70,000
Product Price = $120
```

So:

```text
Numerical Value → Linear Regression
```


## Classification

Classification is used when the target is a **category or class**.

Examples:

```text
Spam / Not Spam
Yes / No
Pass / Fail
Low / Medium / High
```

So:

```text
Category/Class → Classification
```


# Logistic Regression

Even though it has the word "Regression", Logistic Regression is mainly used for **Classification**.

It predicts the **probability of a class**.

Example:

```text
Will the customer buy the product?

Probability = 0.85
```

The model can classify:

```text
0.85 → Buy
0.15 → Not Buy
```

Logistic Regression uses the **Sigmoid Function** to convert the output into a value between 0 and 1.

```text
0 ≤ Probability ≤ 1
```


# Logistic Regression Example

For an e-commerce project:

```text
Customer Information
        ↓
Logistic Regression
        ↓
Probability of Purchase
        ↓
Buy / Not Buy
```
