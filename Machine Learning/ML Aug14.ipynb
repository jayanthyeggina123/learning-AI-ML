# Unsupervised Learning

Unsupervised Learning is a type of machine learning where the data has **no labeled target/output**.

The model tries to find **patterns, groups, or structures** in the data by itself.

Example:

```text
Customer Data
      ↓
No "Buy" or "Not Buy" label
      ↓
Model finds similar customers
      ↓
Customer Groups
```

Common Unsupervised Learning techniques:

- **Clustering**
- **Dimensionality Reduction**


# Clustering

Clustering means **grouping similar data points together**.

Data points in the same cluster should be more similar to each other.

Example:

```text
● ● ●          ○ ○ ○
● ● ●          ○ ○ ○
   Cluster 1      Cluster 2
```

Example in business:

```text
Customers
   ↓
Clustering
   ↓
Budget Customers
Regular Customers
Premium Customers
```

Clustering is an **unsupervised learning technique** because there are no predefined labels.


# K-Means Clustering

K-Means is one of the most commonly used clustering algorithms.

It divides the data into **K clusters**.

Example:

```text
K = 3

Data
 ↓
Cluster 1
Cluster 2
Cluster 3
```

The algorithm uses **centroids** to represent the center of each cluster.


# Centroid

A centroid is the **center point of a cluster**.

Example:

```text
       ●
   ●   ●   ●
       ★
   ●       ●

★ = Centroid
```

The centroid represents the approximate center of the data points in that cluster.


# How K-Means Works

Basic steps:

```text
1. Choose K
      ↓
2. Initialize centroids
      ↓
3. Calculate distance from each data point to centroids
      ↓
4. Assign each point to the nearest centroid
      ↓
5. Recalculate the centroids
      ↓
6. Repeat
      ↓
7. Stop when centroids become stable
```


# Euclidean Distance

Euclidean Distance measures the **straight-line distance between two data points**.

For two points:

```text
A(x₁, y₁)
B(x₂, y₂)
```

Formula:

```text
Distance = √((x₂ - x₁)² + (y₂ - y₁)²)
```

Example:

```text
A = (2, 3)
B = (5, 7)
```

```text
Distance = √((5-2)² + (7-3)²)

Distance = √(9 + 16)

Distance = √25

Distance = 5
```

In K-Means, Euclidean distance is commonly used to find the nearest centroid.


# Assigning Data Points to a Cluster

Suppose we have:

```text
Data Point = P

Centroid 1 → Distance = 2
Centroid 2 → Distance = 7
Centroid 3 → Distance = 4
```

The smallest distance is:

```text
2
```

So:

```text
P → Cluster 1
```

The data point is assigned to the **nearest centroid**.


# Data Points in a Cluster

A cluster contains data points that are relatively similar to each other.

Example:

```text
Cluster 1:

● ● ●
● ● ●
```

The points are close together.

```text
Cluster 2:

○ ○ ○
○ ○ ○
```

The points in Cluster 2 are also close together.

The goal is:

```text
Inside a cluster
→ Points should be close

Between clusters
→ Points should be more separated
```


# Elbow Method

The **Elbow Method** is used to choose a suitable value of **K** in K-Means.

We try different values of K and calculate the **Within-Cluster Sum of Squares (WCSS)**.

Example:

```text
K = 1 → High WCSS
K = 2 → Lower WCSS
K = 3 → Much Lower WCSS
K = 4 → Slightly Lower
K = 5 → Slightly Lower
```

We plot:

```text
K
vs
WCSS
```

The point where the decrease starts slowing down is called the **Elbow Point**.

```text
WCSS
 |
 |\
 | \
 |  \
 |   \
 |    \__
 |       \__
 |____________
       K

       ↑
     Elbow
```


# Why Elbow Method?

Increasing K usually reduces the distance within clusters.

But using too many clusters is not always useful.

The Elbow Method helps find a reasonable balance.

```text
Too Few Clusters
       ↓
Poor grouping

Too Many Clusters
       ↓
Overly complicated

Elbow Point
       ↓
Reasonable K
```


# Silhouette Score

Silhouette Score measures how well each data point fits into its cluster.

It compares:

```text
Distance to its own cluster
        vs
Distance to the nearest other cluster
```

The score ranges from:

```text
-1 to +1
```

Generally:

```text
+1 → Very good clustering
 0 → Overlapping clusters
-1 → Data point may be in the wrong cluster
```

Higher Silhouette Score generally means better-separated and more cohesive clusters.


# Intra-Cluster Distance

**Intra-cluster distance** means the distance between data points **within the same cluster**.

Example:

```text
Cluster 1

● ●
 ● ●
```

If the points are close together:

```text
Low Intra-Cluster Distance
        ↓
Good / Compact Cluster
```

The goal is usually to keep intra-cluster distances **small**.


# Nearest Cluster Distance

Nearest cluster distance measures how close a data point is to the **nearest different cluster**.

Example:

```text
Cluster A       Cluster B

● ● ●           ○ ○ ○
 ● ●             ○ ○
```

If the clusters are far apart:

```text
Large distance between clusters
        ↓
Better separation
```

Silhouette Score uses the idea of:

```text
Own Cluster Distance
        vs
Nearest Other Cluster Distance
```


# Custom Centroid Initialization

In K-Means, the initial centroids can affect the final clusters.

Normally, K-Means can choose initial centroids automatically.

We can also provide our own initial centroid positions.

Example:

```python
import numpy as np
from sklearn.cluster import KMeans

initial_centroids = np.array([
    [2, 3],
    [8, 7],
    [5, 10]
])

model = KMeans(
    n_clusters=3,
    init=initial_centroids,
    n_init=1,
    random_state=42
)
```

Custom initialization can be useful when we already have an idea about where the cluster centers should be.

`n_init=1` is used here because we are providing the initial centroids ourselves.


# Hierarchical Clustering

Hierarchical Clustering creates a hierarchy of clusters.

There are two main approaches:

```text
Agglomerative
Divisive
```

Agglomerative clustering is more commonly used.


# Agglomerative Clustering

Agglomerative clustering starts with **each data point as its own cluster**.

Then it repeatedly combines the closest clusters.

Example:

```text
Start:

A   B   C   D
↓   ↓   ↓   ↓

Each point is its own cluster
```

Then:

```text
A + B
C + D
```

Then:

```text
AB + CD
```

Finally:

```text
ABCD
```

So it works from:

```text
Individual Points
        ↓
Small Clusters
        ↓
Larger Clusters
        ↓
One Large Cluster
```


# Dendrogram

A **Dendrogram** is a tree-like diagram used to visualize hierarchical clustering.

Example:

```text
        ┌───────────────┐
        │               │
      ┌─┴─┐           ┌─┴─┐
      A   B           C   D
```

The height at which clusters are joined represents their distance/dissimilarity.

We can draw a horizontal line across the dendrogram to decide how many clusters we want.

```text
        ───────────────  ← Cut
        |              |
      ┌─┴─┐          ┌─┴─┐
      A   B          C   D
```

This gives:

```text
Cluster 1 → A, B
Cluster 2 → C, D
```


# DBSCAN

DBSCAN stands for:

**Density-Based Spatial Clustering of Applications with Noise**

It is an important clustering algorithm that groups data points based on **density**.

Unlike K-Means, DBSCAN does not require us to specify the number of clusters beforehand.

It can also identify **noise/outliers**.


# Important DBSCAN Parameters

Two important parameters are:

```text
Epsilon (eps)
Min_Points (min_samples)
```


# Epsilon

**Epsilon (`eps`)** defines the maximum distance around a data point to consider other points as its neighbors.

Think of it as a radius.

```text
        ●
     ●  ●  ●
        ●

     ← eps →
```

Points inside this radius are considered neighbors.


# Min_Points

`Min_Points` specifies the minimum number of points required within the epsilon neighborhood for a point to be considered a **Core Point**.

Example:

```text
eps = 2
min_samples = 5
```

If at least 5 points are within the required neighborhood:

```text
→ Core Point
```


# Core Point

A **Core Point** has at least the required number of neighboring points within the specified `eps` distance.

Example:

```text
       ●
    ●  ●  ●
       ●
```

Many points are close together.

```text
Core Point
    ↓
Enough nearby points
```


# Border Point

A **Border Point** does not have enough neighbors to be a core point itself, but it is close enough to a core point.

```text
Core Points
 ● ● ●
 ● ● ●

       ○
       ↑
   Border Point
```

It belongs to the cluster but does not have enough neighbors to become a core point.


# Noise Point

A **Noise Point** does not have enough nearby points and is not close enough to a core point.

It is treated as an outlier/noise.

```text
● ● ●
● ● ●


                  ×
                  ↑
                Noise
```

DBSCAN can therefore identify unusual points automatically.


# DBSCAN Summary

```text
DBSCAN

Core Point
→ Enough nearby points

Border Point
→ Near a core point but does not have enough neighbors

Noise Point
→ Not sufficiently connected to a dense region
```


# K-Means vs DBSCAN

| K-Means | DBSCAN |
|---|---|
| Need to choose K | No need to choose number of clusters |
| Uses centroids | Uses density |
| Works well with roughly spherical clusters | Can find irregular-shaped clusters |
| Sensitive to outliers | Can identify noise/outliers |
| Usually uses distance to centroid | Uses `eps` and `min_samples` |
| Can be affected by initialization | Depends mainly on `eps` and `min_samples` |


# Evaluating Clustering

Since clustering is unsupervised, we usually do not have known target labels.

We can evaluate clustering using:

```text
1. Internal Evaluation
2. External Evaluation
3. Relative Evaluation
```


# Internal Evaluation

Internal evaluation uses only the data and the cluster assignments.

It checks:

```text
How compact are the clusters?
How well separated are the clusters?
```

Examples:

- Silhouette Score
- Davies-Bouldin Index
- Calinski-Harabasz Index


# External Evaluation

External evaluation compares the clusters with **known external labels**.

This is useful when actual labels are available for evaluation, even though the clustering algorithm itself did not use them during training.

Examples:

- Adjusted Rand Index (ARI)
- Normalized Mutual Information (NMI)


# Relative Evaluation

Relative evaluation compares different clustering solutions.

Example:

```text
K = 2 → Silhouette = 0.45
K = 3 → Silhouette = 0.61
K = 4 → Silhouette = 0.52
```

Here:

```text
K = 3
   ↓
Best Silhouette Score
```

So we may prefer the clustering solution with `K = 3`.


# EDA Before Clustering

EDA stands for **Exploratory Data Analysis**.

Before applying a clustering algorithm, we should understand how the data is distributed.

We can check:

- Number of rows and columns
- Missing values
- Data types
- Outliers
- Feature distributions
- Correlations
- Scale of features

Example:

```text
EDA
 ↓
Understand Data
 ↓
Clean Data
 ↓
Scale Features
 ↓
Choose Clustering Algorithm
 ↓
Evaluate Clusters
```


# Why Scaling is Important for Clustering

Distance-based algorithms such as K-Means and DBSCAN are affected by the scale of features.

Example:

```text
Age     → 18 to 60
Income  → 20,000 to 100,000
```

Income has much larger numerical values.

It can dominate distance calculations.

So we often use:

```text
StandardScaler
MinMaxScaler
```

before clustering.


# Which Clustering Model to Use?

The choice depends on the data.

```text
K-Means
→ Simple clustering
→ Roughly spherical groups
→ Need to choose K


Agglomerative
→ Want hierarchical structure
→ Use Dendrogram


DBSCAN
→ Irregular-shaped clusters
→ Many outliers/noise
→ Density-based data
```

A simple approach is:

```text
EDA
 ↓
Understand distribution and outliers
 ↓
Try suitable clustering models
 ↓
Compare evaluation scores
 ↓
Choose the best model
```


# Principal Component Analysis (PCA)

PCA stands for **Principal Component Analysis**.

PCA is a **dimensionality reduction technique**.

It reduces the number of features while trying to preserve as much important information/variation in the data as possible.

Example:

```text
100 Features
     ↓
     PCA
     ↓
10 Features
```

Instead of working with 100 dimensions, we can work with 5 or 10 principal components.


# Why Use PCA?

High-dimensional data can be difficult to:

- Visualize
- Process
- Train models on
- Understand

PCA helps reduce the number of dimensions.

Example:

```text
100 Features
     ↓
PCA
     ↓
10 Principal Components
     ↓
Easier to work with
```


# Principal Components

PCA creates new variables called **Principal Components (PCs)**.

They are combinations of the original features.

```text
Original Features
X₁ X₂ X₃ X₄ X₅
       ↓
      PCA
       ↓
PC1 PC2 PC3
```

The principal components are ordered based on how much variation they explain.

```text
PC1 → Most variation
PC2 → Second most variation
PC3 → Third most variation
```

and so on.


# PCA Example

Suppose a dataset has:

```text
100 Features
```

We may reduce it to:

```text
PC1
PC2
PC3
...
PC10
```

So:

```text
100 Dimensions
       ↓
      PCA
       ↓
10 Dimensions
```

The goal is to retain as much useful information as possible while reducing dimensions.


# PCA and Clustering

PCA can also be useful before clustering when there are many features.

Example:

```text
100 Features
      ↓
Clean Data
      ↓
Scale Data
      ↓
PCA
      ↓
10 Components
      ↓
K-Means / DBSCAN
      ↓
Clusters
```

PCA can also help visualize high-dimensional data by reducing it to 2 or 3 dimensions.

```text
100 Dimensions
      ↓
     PCA
      ↓
2 Dimensions
      ↓
Plot the clusters
```


# Important PCA Point

PCA does **not simply select 5 or 10 original columns**.

It creates **new features called Principal Components** from combinations of the original features.

```text
Original Features
       ↓
      PCA
       ↓
New Principal Components
```
