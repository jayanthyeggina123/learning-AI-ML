# Ensemble Learning

Ensemble Learning means **combining multiple machine learning models** to get a better and more reliable prediction.

Instead of depending on only one model:

```text
One Model
    ↓
Prediction
```

Ensemble learning uses multiple models:

```text
Model 1 ──┐
Model 2 ──┤
Model 3 ──┤
Model 4 ──┘
     ↓
Combined Prediction
     ↓
Better Result
```

The individual models are often called **weak learners** or base models.

Two important Ensemble Learning techniques are:

- **Bagging**
- **Boosting**


# Bagging

Bagging stands for **Bootstrap Aggregating**.

In Bagging, multiple models are trained **independently and in parallel** using different samples of the data.

```text
              Dataset
                 ↓
       ┌─────────┼─────────┐
       ↓         ↓         ↓
    Model 1   Model 2   Model 3
       ↓         ↓         ↓
       └─────────┼─────────┘
                 ↓
        Combined Prediction
```

The models do not depend on each other.

They are trained separately and their results are combined.


# Random Forest

**Random Forest** is a popular Bagging algorithm.

It creates many **Decision Trees** and combines their predictions.

```text
                 Dataset
                    ↓
          ┌─────────┼─────────┐
          ↓         ↓         ↓
       Tree 1     Tree 2     Tree 3
          ↓         ↓         ↓
          └─────────┼─────────┘
                    ↓
             Final Prediction
```

Each tree gets a different random sample of the data and features.

For Classification:

```text
Tree 1 → Cat
Tree 2 → Dog
Tree 3 → Dog
Tree 4 → Dog
Tree 5 → Cat

Final → Dog
```

The majority vote is used.

For Regression:

```text
Tree 1 → $200
Tree 2 → $220
Tree 3 → $210

Final Prediction ≈ Average
```


# Advantages of Random Forest

- Reduces overfitting compared with a single Decision Tree.
- Can handle large datasets.
- Can work with many features.
- Usually gives good prediction performance.
- Can be used for both Classification and Regression.


# Drawbacks of Random Forest

- Uses many Decision Trees, so it can require more computation.
- Can use more memory than a single Decision Tree.
- Training and prediction can be slower when there are many trees.
- The final model is harder to interpret than a single Decision Tree.


# Bagging vs Decision Tree

```text
Decision Tree
    ↓
One Tree
```

```text
Random Forest
    ↓
Many Decision Trees
    ↓
Combine their results
```

So:

```text
Random Forest = Bagging + Decision Trees
```


# Boosting

Boosting is another Ensemble Learning technique.

Unlike Bagging, Boosting trains models **one after another in sequence**.

Each new model tries to improve the mistakes made by the previous models.

```text
Model 1
   ↓
Learn
   ↓
Model 2
   ↓
Correct previous mistakes
   ↓
Model 3
   ↓
Correct previous mistakes
   ↓
Final Prediction
```

So:

```text
Bagging  → Parallel
Boosting → Sequential
```


# AdaBoost

AdaBoost stands for **Adaptive Boosting**.

It trains weak learners one after another.

The observations that were incorrectly predicted receive more importance, so the next model focuses more on those difficult observations.

```text
Dataset
   ↓
Tree 1
   ↓
Find mistakes
   ↓
Increase importance of difficult observations
   ↓
Tree 2
   ↓
Find remaining mistakes
   ↓
Tree 3
   ↓
Final Prediction
```

### Simple Example

Suppose:

```text
Customer 1 → Correct
Customer 2 → Correct
Customer 3 → Wrong
Customer 4 → Correct
```

AdaBoost gives more importance to Customer 3.

The next tree focuses more on the difficult example.

```text
Wrong Prediction
       ↓
Increase Weight
       ↓
Next Tree Pays More Attention
```


# AdaBoost and Weights

AdaBoost works by assigning **weights** to training observations.

```text
Initial Data
     ↓
Equal Weights
     ↓
Train Model
     ↓
Find Wrong Predictions
     ↓
Increase Their Weights
     ↓
Train Next Model
```

The process continues until the required number of weak learners is created.

The final prediction combines the predictions from the different learners.


# Gradient Boosting

Gradient Boosting also builds models **sequentially**.

Each new model tries to reduce the errors made by the previous models.

Usually, Decision Trees are used as the weak learners.

```text
Tree 1
   ↓
Calculate Error
   ↓
Tree 2 learns from Error
   ↓
Calculate New Error
   ↓
Tree 3 learns from Error
   ↓
Final Prediction
```

The goal is to gradually reduce the model's **loss/cost function**.


# Gradient Boosting and Gradient Descent

Gradient Boosting is related to the idea of **Gradient Descent**.

Gradient Descent tries to reduce the cost/loss:

```text
High Cost
    ↓
Gradient Descent
    ↓
Lower Cost
    ↓
Better Model
```

Gradient Boosting uses this idea to build new learners that improve the previous model's errors.


# XGBoost

**XGBoost** stands for **Extreme Gradient Boosting**.

It is an optimized and powerful implementation of Gradient Boosting.

```text
Gradient Boosting
       ↓
XGBoost
       ↓
Optimized / Faster Implementation
```

XGBoost is widely used for:

- Classification
- Regression
- Structured/tabular data
- Large datasets


# Advantages of XGBoost

- High prediction performance.
- Handles structured/tabular data very well.
- Includes regularization to help control overfitting.
- Can handle missing values.
- Often performs well in machine learning competitions and practical projects.


# CatBoost

**CatBoost** is another Gradient Boosting algorithm.

It is especially useful when the dataset contains many **categorical features**.

Example:

```text
Brand
City
Gender
Category
```

CatBoost can handle categorical features effectively with less manual encoding compared with many traditional approaches.

```text
Categorical Data
       ↓
CatBoost
       ↓
Gradient Boosting
       ↓
Prediction
```


# LightGBM

**LightGBM** stands for **Light Gradient Boosting Machine**.

It is a fast and memory-efficient Gradient Boosting framework developed by Microsoft.

It is designed to work well with **large/high-volume datasets**.

```text
Large Dataset
     ↓
LightGBM
     ↓
Fast Training
     ↓
Prediction
```

It is especially useful when:

- Dataset is large
- Number of features is large
- Training speed is important


# XGBoost vs LightGBM vs CatBoost

| Model | Main Advantage |
|---|---|
| **XGBoost** | Strong performance and regularization |
| **LightGBM** | Fast and efficient for large datasets |
| **CatBoost** | Very good with categorical features |


# Bagging vs Boosting

| Bagging | Boosting |
|---|---|
| Models are trained in parallel | Models are trained sequentially |
| Models are mostly independent | Each model depends on previous results |
| Reduces variance | Mainly reduces bias and improves weak learners |
| Random Forest is an example | AdaBoost, Gradient Boosting, XGBoost are examples |
| Usually easier to parallelize | Sequential nature makes it less parallel |


# AdaBoost vs Gradient Boosting

### AdaBoost

```text
Model 1
   ↓
Increase weights of wrong observations
   ↓
Model 2
   ↓
Increase weights again
   ↓
Model 3
```

### Gradient Boosting

```text
Model 1
   ↓
Calculate errors/residuals
   ↓
Model 2 learns from errors
   ↓
Model 3 learns from new errors
```

Simple difference:

```text
AdaBoost → Focuses on difficult observations using weights

Gradient Boosting → Builds new models to reduce the loss/errors
```


# Regression Metrics

When we use Regression models, we need metrics to check how well the model performs.

Important regression metrics include:

- **MAE**
- **MSE**
- **RMSE**
- **R²**
- **Adjusted R²**

## MAE

**Mean Absolute Error**

It calculates the average absolute difference between actual and predicted values.

```text
MAE = Average |Actual - Predicted|
```

Lower MAE is better.

Example:

```text
Actual     = 100
Predicted  = 90

Error = 10
```


## MSE

**Mean Squared Error**

It squares the errors before taking the average.

```text
MSE = Average (Actual - Predicted)²
```

Large errors are penalized more because they are squared.

Lower MSE is better.


## RMSE

**Root Mean Squared Error**

It is the square root of MSE.

```text
RMSE = √MSE
```

Lower RMSE is better.

RMSE is useful because it is in the **same unit as the target variable**.


## R²

R² tells us how much of the variation in the target variable is explained by the model.

```text
R² = 1
```

means the model explains the data very well.

Generally:

```text
Higher R² → Better model fit
```


# KNN

**KNN** stands for **K-Nearest Neighbors**.

It is a supervised learning algorithm.

It predicts a new observation based on the nearest observations in the training data.

```text
New Data Point
      ↓
Find K nearest points
      ↓
Look at their classes/values
      ↓
Make Prediction
```

For Classification:

```text
Nearest neighbors:
Dog
Dog
Cat

Prediction → Dog
```

For Regression:

```text
Nearest values:
100
120
110

Prediction ≈ 110
```


# SVM

**SVM** stands for **Support Vector Machine**.

It is a supervised learning algorithm mainly used for classification.

It tries to find the best **boundary/hyperplane** that separates different classes.

```text
Class A       |       Class B
● ● ● ●       |       ○ ○ ○ ○
● ● ●         |       ○ ○ ○
              ↑
        Decision Boundary
```

The goal is to create a boundary with the **largest possible margin** between the classes.

SVM can also be used for regression, called **SVR (Support Vector Regression)**.
