# Logistic Regression

Logistic Regression is a **classification algorithm** used to predict which class a particular value belongs to.

It is mainly used when the output is a **category or class**.

Example:

```text
Email → Spam / Not Spam
Customer → Buy / Not Buy
Student → Pass / Fail
```

The main idea is:

```text
Particular value → Which class does it belong to?
```



# Logistic Regression and Linear Regression

Logistic Regression is similar to Linear Regression because it starts with a linear equation:

```text
Y = mx + c
```

But Logistic Regression does not directly give the final class.

It uses the linear output and passes it through a **Sigmoid Function** to get a probability.

```text
Input
  ↓
Linear Equation
  ↓
Sigmoid Function
  ↓
Probability
  ↓
Class
```



# Boundary Line

A **decision boundary** is used to separate different classes.

Example:

```text
Class 0 | Class 1
        |
        |  Decision Boundary
        |
```

The model uses the boundary to decide which class a particular data point belongs to.



# Sigmoid Function

The **Sigmoid Function** converts the model output into a probability between **0 and 1**.

```text
0 ≤ Probability ≤ 1
```

The output can then be used to classify the data.

Example:

```text
Probability = 0.80
```

This means the model is giving an **80% probability** for a particular class.



# Binary Classification

When there are only **two classes**, it is called Binary Classification.

Examples:

```text
Spam / Not Spam
Yes / No
Buy / Not Buy
Pass / Fail
```

The Sigmoid Function is commonly used for binary classification.



# Threshold Value

A **threshold** is the value used to convert probability into a class.

The common threshold is:

```text
0.5
```

Example:

```text
Probability < 0.5 → Class 0
Probability ≥ 0.5 → Class 1
```

For example:

```text
Probability = 0.30
→ Class 0

Probability = 0.80
→ Class 1
```

The threshold can sometimes be changed depending on the problem.



# Softmax Function

The **Softmax Function** is commonly used when there are **more than two classes**.

It converts the model outputs into probabilities for each class.

Example:

```text
Product Category:

Shirt      → 0.20
Jeans      → 0.55
Jacket     → 0.25
```

The probabilities add up to:

```text
0.20 + 0.55 + 0.25 = 1.00
```

The class with the highest probability is usually selected.

```text
Jeans → 0.55 → Predicted Class
```



# Multinomial Logistic Regression

Multinomial Logistic Regression is used when there are **more than two classes**.

Example:

```text
Product Type:

T-Shirt
Jeans
Jacket
```

Instead of predicting only Yes/No, the model predicts one class from multiple categories.

```text
Input
  ↓
Multinomial Logistic Regression
  ↓
Probabilities for each class
  ↓
Class with highest probability
```



# Classification Metrics

Classification metrics are used to measure **how well a classification model performs**.

Some common classification metrics are:

- Accuracy
- Precision
- Recall
- F1 Score
- Specificity
- ROC Curve
- AUC


# Confusion Matrix

A **Confusion Matrix** compares the **Actual Values** with the **Predicted Values**.

Basic structure:

|  | Actual Positive | Actual Negative |
|---|---|---|
| **Predicted Positive** | TP | FP |
| **Predicted Negative** | FN | TN |



# TP - True Positive

The model predicted **Positive**, and the actual value was also **Positive**.

Example:

```text
Actual = Spam
Predicted = Spam
```

This is a **True Positive (TP)**.



# TN - True Negative

The model predicted **Negative**, and the actual value was also **Negative**.

Example:

```text
Actual = Not Spam
Predicted = Not Spam
```

This is a **True Negative (TN)**.



# FP - False Positive

The model predicted **Positive**, but the actual value was **Negative**.

Example:

```text
Actual = Not Spam
Predicted = Spam
```

This is a **False Positive (FP)**.

It is also called a **Type 1 Error**.



# FN - False Negative

The model predicted **Negative**, but the actual value was **Positive**.

Example:

```text
Actual = Spam
Predicted = Not Spam
```

This is a **False Negative (FN)**.

It is also called a **Type 2 Error**.



# Easy Way to Remember TP, TN, FP, FN

```text
TP → Predicted Positive + Actually Positive
TN → Predicted Negative + Actually Negative
FP → Predicted Positive + Actually Negative
FN → Predicted Negative + Actually Positive
```


# Accuracy

Accuracy tells us how many predictions the model classified correctly out of all predictions.

Formula:

```text
Accuracy = (TP + TN) / (TP + TN + FP + FN)
```

Example:

```text
100 predictions
90 were correct

Accuracy = 90 / 100
         = 90%
```

Accuracy works well when the classes are reasonably balanced.


# Recall / Sensitivity

Recall tells us how many of the **actual positive cases** were correctly identified by the model.

Formula:

```text
Recall = TP / (TP + FN)
```

Example:

If there are 100 actual positive cases and the model correctly identifies 90:

```text
Recall = 90 / 100
       = 90%
```

Recall is important when **missing a positive case is a serious problem**.

Example:

```text
Disease Detection
```

We want to identify as many actual positive patients as possible.


# Precision

Precision tells us how many of the cases predicted as **positive were actually positive**.

Formula:

```text
Precision = TP / (TP + FP)
```

Example:

The model predicts 100 people as positive.

Out of them, 90 are actually positive.

```text
Precision = 90 / 100
          = 90%
```

Precision is important when **false positives are a serious problem**.


# Precision and Recall Trade-off

There can be a **trade-off between Precision and Recall**.

Increasing the threshold can sometimes:

```text
Precision ↑
Recall ↓
```

Decreasing the threshold can sometimes:

```text
Recall ↑
Precision ↓
```

So we need to choose the threshold depending on the problem.

Example:

### Medical Diagnosis

We may prefer higher **Recall** because missing a sick patient can be dangerous.

### Spam Detection

We may prefer higher **Precision** because we don't want important emails to be incorrectly classified as spam.


# F1 Score

F1 Score combines **Precision and Recall** into one metric.

Formula:

```text
F1 Score = 2 × (Precision × Recall) / (Precision + Recall)
```

It is useful when we want a balance between Precision and Recall.

F1 Score ranges from:

```text
0 to 1
```

Higher F1 Score generally means better performance.


# Specificity

Specificity tells us how well the model identifies **actual negative cases**.

Formula:

```text
Specificity = TN / (TN + FP)
```

Example:

If there are 100 actual negative cases and the model correctly identifies 90:

```text
Specificity = 90 / 100
            = 90%
```

Specificity is also called the **True Negative Rate**.


# Recall vs Specificity

| Metric | Measures |
|---|---|
| **Recall / Sensitivity** | How many actual positives were correctly identified |
| **Specificity** | How many actual negatives were correctly identified |

```text
Recall      → Focuses on Positive cases
Specificity → Focuses on Negative cases
```


# ROC Curve

ROC stands for:

**Receiver Operating Characteristic**

The **ROC Curve** is used to evaluate the performance of a binary classification model at different threshold values.

It shows the relationship between:

```text
True Positive Rate (TPR)
```

and

```text
False Positive Rate (FPR)
```

Recall is the same as True Positive Rate:

```text
TPR = TP / (TP + FN)
```

False Positive Rate:

```text
FPR = FP / (FP + TN)
```


# ROC Curve - Simple Idea

The classification threshold can be changed.

For example:

```text
Threshold = 0.3
Threshold = 0.5
Threshold = 0.7
```

For each threshold, the model will have different:

- True Positive Rate
- False Positive Rate

The ROC Curve shows these values on a graph.


# AUC

AUC stands for:

**Area Under the Curve**

It represents the **area under the ROC curve**.

AUC is generally between:

```text
0 and 1
```

Higher AUC generally means the model is better at distinguishing between the classes.

Example:

```text
AUC = 0.95 → Very good
AUC = 0.80 → Good
AUC = 0.50 → Random guessing
```



# ROC Curve and AUC

```text
ROC → Curve
AUC → Area under the ROC Curve
```

The area under each ROC curve can be compared to understand which classification model performs better.

# Decision Tree

A **Decision Tree** is a supervised machine learning algorithm that can be used for **classification and regression**.

It makes decisions by asking a series of questions about the features.

Example:

```text
Is price > $100?
       ↓
      Yes
       ↓
Is rating > 4?
   ↓         ↓
 Yes        No
  ↓          ↓
Buy        Not Buy
```

The tree contains:

- **Root Node** → Starting point
- **Decision Nodes** → Questions/conditions
- **Branches** → Possible outcomes
- **Leaf Nodes** → Final prediction
