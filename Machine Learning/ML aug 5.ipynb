# Machine Learning

Machine Learning (ML) is a branch of **Artificial Intelligence (AI)** that enables computers to **learn from data** and make predictions or decisions **without being explicitly programmed for every task**.

It learns from **past data** to make predictions or decisions on **new data**.



### Examples

- Emails marked as **Spam** and **Not Spam**.
- Netflix recommends movies based on your watch history.
- Amazon recommends products you may want to buy.



### Machine Learning Process

1. Collect data.
2. Clean and prepare the data.
3. Choose a Machine Learning algorithm.
4. Train the model using the data.
5. Test the model with new data.
6. Use the trained model to make predictions.



### Advantages of Machine Learning

- Automates repetitive tasks.
- Handles very large amounts of data.
- Improves prediction accuracy.


### Limitations of Machine Learning

- Requires a large amount of quality data.
- Training can take a long time.
- Performance depends on the quality of the data.






# Machine Learning Lifecycle

###  Data Description

Understand the dataset by examining:

- Features (columns)
- Data types
- Number of records
- Missing values
- Relationships between variables

This step tells the story of the data and helps understand what information is available.



###  Statistical and Graphical Data Analysis (EDA)

Perform **Exploratory Data Analysis (EDA)** to understand the data.

This includes:

- Summary statistics
- Histograms
- Bar charts
- Box plots
- Correlation analysis
- Distribution analysis

The goal is to identify patterns, trends, missing values, and outliers.



###  Data Transformation

Prepare the data before training the model.

This may include:

- Handling missing values
- Encoding categorical variables
- Creating new features
- Removing unnecessary columns
- Feature engineering



###  Selection of a Machine Learning Algorithm

Choose a suitable machine learning algorithm based on the problem and the patterns observed during EDA.

Examples:

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- K-Nearest Neighbors (KNN)
- Support Vector Machine (SVM)



###  Data Standardization and Normalization

Scale the numerical data so that all features are on a similar range.

Common techniques include:

- Standardization
- Normalization

This improves the performance of many machine learning algorithms.



###  Creation of Training and Testing Datasets

Split the dataset into:

- **Training Data** – Used to train the model.
- **Testing Data** – Used to evaluate the model.

A common split is:

- 80% Training
- 20% Testing



###  Model Training

Train the selected machine learning algorithm using the training dataset.

During training, the model learns patterns and relationships from the data.



###  Model Accuracy Evaluation

Evaluate how well the model performs.

Calculate:

- Training Accuracy
- Testing Accuracy

Other evaluation metrics may include:

- Mean Squared Error (MSE)
- R² Score
- Precision
- Recall
- F1-Score

The evaluation method depends on the type of machine learning problem.



###  Hyperparameter Tuning

Adjust the model's parameters to improve performance and accuracy.

Examples include changing:

- Learning rate
- Maximum tree depth
- Number of neighbors
- Number of estimators

The goal is to build the best-performing model.



###  Saving the Model

Save the trained machine learning model for future use.

Common Python libraries:

- Pickle
- Joblib

The saved model can later be loaded without retraining.



###  Model Deployment

Deploy the trained model so that users or applications can use it.

Common deployment methods include:

- Live Stream
- Batch Processing
- Mini-Batch Processing



###  Production Deployment and Testing

Deploy the model into the production environment.

Test it with real-world data to ensure it performs correctly and reliably.



###  Retraining the Model

As new data becomes available, retrain the model to improve its performance and maintain accuracy.

This helps the model adapt to changing data patterns.



###  Logging and Monitoring

Continuously monitor the deployed model.

Track:

- Predictions
- Errors
- Performance
- Accuracy
- System logs

Monitoring helps detect problems and maintain the model over time.



###  Dashboard for Monitoring and Reports

Create dashboards to visualize:

- Model accuracy
- Predictions
- Errors
- Performance metrics
- Logs

Common tools include:

- Streamlit
- Power BI
- Tableau
- Grafana

Dashboards help monitor the model in real time.



###  Stop

The machine learning lifecycle is complete.

However, in real-world projects, the process is continuous because models are regularly monitored, updated, and retrained as new data becomes available.

# Comparison of Machine Learning Types

| Type | Training Data | Output | Example |
|------|---------------|--------|---------|
| **Supervised Learning** | Labeled data | Predicts numerical values or categories | House price prediction |
| **Regression** | Labeled data | Numerical value | Salary prediction |
| **Classification** | Labeled data | Category / Class | Spam detection |
| **Unsupervised Learning** | Unlabeled data | Finds groups or patterns | Customer segmentation |
| **Reinforcement Learning** | Rewards and penalties | Learns the best action | Self-driving cars |
| **Semi-Supervised Learning** | Small amount of labeled data + large amount of unlabeled data | Improves prediction accuracy | Medical image classification |



- **Supervised Learning** → Patterns (correct answers/labels) are given.
- **Regression** → Predicts a **number** (continuous value).
- **Classification** → Predicts a **category or class**.
- **Unsupervised Learning** → Finds **patterns or groups by itself** (no labels are given).
- **Reinforcement Learning** → Learns through **rewards and punishments**.
- **Semi-Supervised Learning** → Uses a **small amount of labeled data** and a **large amount of unlabeled data**.

# Linear Regression Equation
Y=mX+c
	​


Where:

Y = Dependent Variable (Target/Output)
X = Independent Variable (Feature/Input)
m = Slope (Coefficient)
c = Intercept (Constant)

This equation is used to predict the value of Y based on X.


# Multiple Linear Regression

Multiple Linear Regression is a supervised machine learning algorithm that uses **two or more independent variables (features)** to predict **one dependent variable (target)**.

Unlike Simple Linear Regression, which uses only one input variable, Multiple Linear Regression considers multiple features to improve prediction accuracy.



## Equation

```text
Y = b₀ + b₁X₁ + b₂X₂ + b₃X₃ + ... + bₙXₙ
```



## Where:

- **Y** = Dependent Variable (Target/Output)
- **b₀** = Intercept (Constant)
- **b₁, b₂, b₃, ..., bₙ** = Coefficients (Weights) of each feature
- **X₁, X₂, X₃, ..., Xₙ** = Independent Variables (Input Features)




# Difference Between R, R², and Adjusted R²

| Measure | Meaning | Range | Best Used For |
|---------|---------|-------|---------------|
| **R** | Measures the strength and direction of the relationship between variables. | **-1 to +1** | Correlation |
| **R² (R Square)** | Measures how well the regression model explains the variation in the data. | **0 to 1 (0%–100%)** | Evaluating the fit of a regression model |
| **Adjusted R²** | Measures model fit while accounting for the number of predictors (independent variables). | **Usually 0 to 1** | Multiple Linear Regression |

# Gradient Descent

Gradient Descent is an **optimization algorithm** used to find the **best values of the model parameters (weights and bias)** by minimizing the **cost (loss) function**.

Its goal is to reduce the prediction error and improve the model's accuracy.



## Why do we need Gradient Descent?

When a machine learning model makes predictions, there is usually some error.

Gradient Descent repeatedly adjusts the model parameters until the error becomes as small as possible.



## How it Works

1. Start with random values of weights.
2. Calculate the prediction error (cost).
3. Calculate the gradient (direction of steepest increase).
4. Move in the opposite direction of the gradient.
5. Repeat until the cost reaches its minimum.



## Example

Suppose you are predicting house prices.

```text
Predicted Price = $250,000
Actual Price = $220,000
```

The prediction has an error.

Gradient Descent changes the model's parameters little by little until the prediction becomes closer to the actual value.

# Optimizer

An **Optimizer** is the algorithm that updates the model's parameters (weights and bias) to reduce the cost (loss).

Gradient Descent is one of the most common optimizers.



## Purpose of an Optimizer

- Reduce the cost function.
- Improve prediction accuracy.
- Find the best model parameters.



## Common Optimizers

- Gradient Descent (GD)
- Stochastic Gradient Descent (SGD)
- Mini-Batch Gradient Descent
- Adam
- RMSProp



# Cost Function (Loss Function)

The **Cost Function** measures how far the model's predictions are from the actual values.

A smaller cost means the model performs better.

The objective of machine learning is to minimize the cost function.



## Example

```text
Actual Marks = 90
Predicted Marks = 70
```

Since the prediction is far from the actual value, the cost is high.

Gradient Descent updates the model until the prediction becomes closer to 90.


# Local Minimum

A **Local Minimum** is a point where the cost is lower than the nearby values, but it is **not the lowest possible cost**.

The optimizer may get stuck at this point.

```text
Cost
 ^
 |          __
 |        _/  \_
 |      _/      \__
 |_____/___________\____
        Local
       Minimum
```



# Global Minimum

A **Global Minimum** is the **lowest possible value of the cost function**.

This is the ideal solution because it gives the best model performance.

```text
Cost
 ^
 |      __
 |    _/  \_
 |   /      \__
 |__/__________\________
            Global
           Minimum
```



# Difference Between Local Minimum and Global Minimum

| Local Minimum | Global Minimum |
|---------------|----------------|
| Lowest point in a small region | Lowest point in the entire graph |
| Model may get stuck here | Best possible solution |
| Error is low | Error is the lowest possible |

# TSS (Total Sum of Squares)

TSS measures the **total variation** in the dependent variable.

It represents how much the actual values vary from their mean.

## Formula

```text
TSS = Σ(yi − ȳ)²
```

Where:

- **yi** = Actual value
- **ȳ** = Mean of actual values



# RSS (Residual Sum of Squares)

RSS measures the **unexplained variation** or prediction error.

A smaller RSS means the model fits the data better.

## Formula

```text
RSS = Σ(yi − ŷi)²
```

Where:

- **yi** = Actual value
- **ŷi** = Predicted value



# ESS (Explained Sum of Squares)

ESS measures the variation explained by the regression model.

A larger ESS means the model explains more of the variation in the data.

## Formula

```text
ESS = Σ(ŷi − ȳ)²
```

Where:

- **ŷi** = Predicted value
- **ȳ** = Mean of actual values



# Relationship Between TSS, ESS, and RSS

The total variation in the data is divided into:

- Variation explained by the model (ESS)
- Variation not explained by the model (RSS)

## Formula

```text
TSS = ESS + RSS
```



## Example

```text
TSS = 100
ESS = 85
RSS = 15
```

```text
TSS = ESS + RSS
100 = 85 + 15


This means:

- The model explains **85%** of the variation.
- The remaining **15%** is unexplained error.
