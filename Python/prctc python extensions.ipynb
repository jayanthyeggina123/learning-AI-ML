```python
# Python extensions :
# File Operations : used to create, read, write, update, and delete files in Python.
# uses the built-in open() function to work with files.
file = open("filename.txt", "mode")
#"r" -	Read a file
#"w" -	Write to a file (creates a new file or overwrites an existing one)
#"a"	- Append data at the end of a file
#"r+"-	Read and write
#"x" -	Create a new file (fails if it already exists)
```

```python
#Working with Text Files
#Text files store data as plain text, such as .txt files.
# read mode :
file = open("students.txt", "r")

print(file.read())

file.close()
```

```python
#Read First 10 Characters
#read(n) reads only the first n characters.
file = open("students.txt", "r")

print(file.read(10))

file.close()
```

```python
#Read Line by Line
#readline() reads one line at a time.
file = open("students.txt", "r")

print(file.readline())
print(file.readline())

file.close()
#readlines() reads all lines and stores them in a list.
```

```python
#Read File into a List
#readlines() reads all lines and stores them in a list.
file = open("students.txt", "r")

data = file.readlines()

print(data)

file.close()
#Each line becomes one element in the list.
```

```python
#Write Mode (w)
#Writes data to a file.
#If the file exists, its previous contents are erased. If the file doesn't exist, Python creates it.
file = open("students.txt", "w")

file.write("John")

file.close()
#The file now contains only:John. because "w" overwrites the existing contents.
```

```python
#Append Mode (a)
#Adds new data to the end of the file without deleting existing data.
file = open("students.txt", "a")

file.write("\nAlice")

file.close()
```

```python
#R+ mode : Allows both reading and writing in the same file.
file = open("students.txt", "r+")

print(file.read())

file.write("\nBob")

file.close()
#Reads the file.Then writes "Bob" into the same file.
```

```python
# copy a file : 
source = open("students.txt", "r")

data = source.read()

destination = open("copy.txt", "w")

destination.write(data)

source.close()
destination.close()
#Reads data from students.txt.Writes it into copy.txt.
```

```python
# Exception Handling : prevents a program from crashing when an error occurs.

#try and except : try contains code that might cause an error. except runs if an error occurs.
try:
    number = int(input("Enter a number: "))
    print(number)

except:
    print("Invalid Input")
# If the user enters text instead of a number, the program prints "Invalid Input" instead of crashing.


# try, except, finally : finally always executes, whether an exception occurs or not.
try:
    print(10/0)

except:
    print("Cannot divide by zero")

finally:
    print("Program Finished")

# Division by zero causes an exception.except handles it.finally still runs.

#try and finally : 
#finally executes even if there is no except block.This is commonly used to close files or release resources.
file = open("students.txt", "r")

try:
    print(file.read())

finally:
    file.close()
    print("File Closed")
#The file is always closed, even if an error occurs.
```

```python
#Connecting to Databases (SQLite)
#SQLite is a lightweight database that comes built into Python.
#It stores all the data in a single database file, so you don't need to install a separate database server.
# import sqlite
import sqlite3

# create database 
import sqlite3
conn = sqlite3.connect("students.db")
print("Database Connected")
#Creates a database named students.db.If it already exists, Python opens it.

# create table 
import sqlite3

conn = sqlite3.connect("students.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS student(
id INTEGER PRIMARY KEY,
name TEXT
)
""")

conn.commit()

print("Table Created")

conn.close()

#connect() opens the database.
#cursor() executes SQL commands.
#CREATE TABLE creates the table.
#commit() saves the changes.
#close() closes the connection.
```
