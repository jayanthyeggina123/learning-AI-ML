```python
#A function is a block of reusable code that performs a specific task. It helps avoid writing the same code multiple times.
def greet():
    print("Hello")

greet()
#def is used to create a function.
#greet() calls (executes) the function.
#The code inside the function runs only when it is called.
```

```python
#fun using return : return statement sends a value back to the place where the function was called.
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
#return a + b returns the sum.The returned value is stored in result.
#print(result) displays the output.
```

```python
# adding fun :
def add_num(a, b):
    return a + b

print(add_num(5, 7))
#The function takes two numbers.Adds them together.Returns the result.
```

```python
#is_prime() Function : Checks whether a number is prime or not.
def is_prime(num):

    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True

print(is_prime(7))
#If the number is less than 2, it is not prime.Check divisibility from 2 to num-1.
#If divisible, return False.
```

```python
# Iterators : An iterator is an object that returns one value at a time. 
# It keeps track of the current position and does not return all values at once.
numbers = [10, 20, 30]

it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))

#iter() creates an iterator.
#next() returns the next value.for loops use iterators internally.
```

```python
# generators : is a special function that uses yield to return one value at a time instead of returning all values at once.
def function_name():
    yield value
```

```python
# gencubes(n) :
def genCubes(n):

    for i in range(1, n + 1):
        yield i ** 3

for value in genCubes(5):
    print(value)
#yield returns one cube at a time. The function pauses after each yield.
#It continues from where it stopped when called again.
```

```python
# genFibon(): 
def genFibon(n):

    a = 0
    b = 1

    for i in range(n):
        yield a
        a, b = b, a + b

for num in genFibon(6):
    print(num)
#Generates Fibonacci numbers one by one.
#yield returns each number without storing the whole sequence.
```

```python
# map() : function applies a function to every element in an iterable and returns a map object.
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

print(result)
#map() applies the function to every element.
#list() converts the result into a list.
```

```python
# reduce() : function takes multiple values and reduces them to a single value.
from functools import reduce

numbers = [1, 2, 3, 4]

result = reduce(lambda a, b: a + b, numbers)

print(result)
#Adds all the numbers together.Returns one final value.
```

```python
# Lambda Function : lambda function is a small anonymous function written in a single line.
lambda arguments : expression

square = lambda x: x * x

print(square(5))
#No def keyword is used.Used for short, simple functions.
#Commonly used with map(), filter(), and reduce().
```

```python
# filter() : function returns only the elements that satisfy a given condition.
numbers = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)

#Checks each element.Keeps only the elements where the condition is True.
#Returns the filtered values.
```

```python
Function:	Reuse a block of code
return:	Sends a value back from a function
add_num():Adds two numbers
is_prime():	Checks if a number is prime
Iterator:	Returns one value at a time using iter() and next()
Generator:	Uses yield to generate values one at a time
genCubes():	Generates cube numbers
genFibon():	Generates Fibonacci numbers
map():	Applies a function to every element
reduce():	Combines multiple values into one value
lambda::	Returns only the values that satisfy a condition
```
